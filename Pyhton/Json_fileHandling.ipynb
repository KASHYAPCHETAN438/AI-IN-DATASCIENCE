# 🧠 What is JSON?

- JSON (JavaScript Object Notation) is a lightweight format used to store and exchange data between a client and a server.

- It is language-independent and very similar to Python dictionaries.

<br>

#### 🧩 Example of JSON
    {
    "name": "Mohit",
    "age": 22,
    "skills": ["Python", "React", "Spring Boot"],
    "isStudent": true
    }




## Why Use JSON in Python?

- To store structured data (like dictionaries/lists).

- To share data between applications.

- To save and read data from files easily.

- Commonly used in APIs and web applications.

## Import JSON Module

```python
import json
```

### The json module provides functions to:

- Convert Python objects to JSON

- Convert JSON data back to Python

- Read from or write to JSON files

## Onversion Between Python and JSON

| Python Type | JSON Equivalent |
| ----------- | --------------- |
| dict        | object          |
| list        | array           |
| str         | string          |
| int / float | number          |
| True        | true            |
| False       | false           |
| None        | null            |


<br>

## 📦 Convert Python → JSON

-  Using json.dumps()

- Converts Python object into a JSON string.

```python
import json

data = {"name": "Mohit", "age": 22, "city": "Delhi"}
json_string = json.dumps(data)

print(json_string)
```

```python
data = {"name": "Mohit", "age": 22, "city": "Delhi"}

with open("data.json", "w") as file:
    json.dump(data, file)

print(data)
```

<br>

## 📥  Convert JSON → Python

-  Using json.loads()

- Converts JSON string into Python object.



```python
json_data = '{"name": "Mohit", "age": 22, "city": "Delhi"}'
python_data = json.loads(json_data)

print(python_data)
print(python_data["name"])
```

```python
with open("data.json", "r") as file:
    data = json.load(file)

print(data)
```

## 🧰  Writing and Reading JSON Files

- Write to JSON File


```python
student = {
    "name": "Chetan",
    "age": 23,
    "course": "B.Tech"
}

with open("student.json", "w") as f:
    json.dump(student, f, indent=4)
```

## Read from JSON File


```python
with open("student.json", "r") as f:
    data = json.load(f)

print(data["name"])   # Output: Chetan
```

## 🧩 4. Parameters of json.dump() and json.dumps()

| Parameter      | Description                        |
| -------------- | ---------------------------------- |
| `indent`       | Adds indentation for pretty output |
| `sort_keys`    | Sorts keys alphabetically          |
| `separators`   | Changes item separators            |
| `ensure_ascii` | Allows non-ASCII characters        |


##  Summary Table

| Function               | Description                         |
| ---------------------- | ----------------------------------- |
| `json.dumps(obj)`      | Convert Python → JSON string        |
| `json.dump(obj, file)` | Write JSON to file                  |
| `json.loads(str)`      | Convert JSON string → Python object |
| `json.load(file)`      | Read JSON file → Python object      |
