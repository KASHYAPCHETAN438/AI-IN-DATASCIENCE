# ✅  What is a Module?
- A **module** in Python is a file containing Python code (functions, variables, classes, etc.).
- Any `.py` file is considered a module.

<br>

### Characterstits
-  Code reusability  
-  Code organization  
-  Makes code readable and maintainable

<br>

## Types of Modules

| Type                            | Description               | Examples                      |
| ------------------------------- | ------------------------- | ----------------------------- |
| **Standard / Built-in Modules** | Pre-installed with Python | `math`, `os`, `sys`, `random` |
| **User-defined Modules**        | Created by the programmer | `calculator.py`               |
| **Third-party Modules**         | Installed via pip         | `numpy`, `pandas`, `django`   |

## 3. platform

- Purpose: get info about OS/platform.

- Example:

```python
# example_platform.py
import platform

print("Platform:", platform.system())        # 'Windows', 'Linux', 'Darwin'
print("Platform info:", platform.platform())
print("Processor:", platform.processor())
```

## 4. subprocess

- Purpose: run external programs, capture output.

- Example:

```python
# example_subprocess.py
import subprocess

# runs 'echo Hello' (on Windows use ['cmd','/c','echo','Hello'])
result = subprocess.run(['echo', 'Hello from subprocess'], capture_output=True, text=True)
print("Return code:", result.returncode)
print("Output:", result.stdout.strip())
```

## 1. Date Module

- The datetime module is used to work with date and time — creation, formatting, arithmetic, and more.

<br>

#### Importing the Module



```python
import datetime
```

#### 🧱 I) Getting Current Date and Time

```python
# Current date and time
now = datetime.datetime.now()
print("Current Date & Time:", now)

# Current date only
today = datetime.date.today()
print("\nToday's Date:", today)
```

<br>

#### 🧩 II) Create a Specific Date or Time

```python
# Create a specific date
d = datetime.date(2025, 10, 30)      # format must be YY-MM-DD
print("Specific Date:", d)

# Create a specific time
t = datetime.time(14, 45, 30)
print("\nSpecific Time:", t)

# Combine date and time
dt = datetime.datetime(2025, 10, 30, 14, 45, 30)
print("\nDateTime Object:", dt)
```

<br>

#### 🕰️ III) Extracting Components

```python

now = datetime.datetime.now()

print("Year:", now.year)
print("Month:", now.month)
print("Day:", now.day)
print("Hour:", now.hour)
print("Minute:", now.minute)
print("Second:", now.second)
```

<br>

#### 🧮 VI) Date Arithmetic (Add / Subtract Days)

```python
from datetime import date, timedelta

today = date.today()
print("Today:", today)

# Add 5 days
future = today + timedelta(days=5)
print("After 5 Days:", future)

# Subtract 10 days
past = today - timedelta(days=10)
print("10 Days Ago:", past)
```

<br>

#### 🧾 V) Formatting Date and Time (strftime)

<br>

**Common Format Codes**

| Code | Meaning             | Example  |
| ---- | ------------------- | -------- |
| `%d` | Day (01–31)         | 30       |
| `%m` | Month (01–12)       | 10       |
| `%Y` | Year (4-digit)      | 2025     |
| `%H` | Hour (00–23)        | 14       |
| `%M` | Minute (00–59)      | 45       |
| `%S` | Second (00–59)      | 30       |
| `%A` | Weekday (Full name) | Thursday |

<br>


```python
import datetime

now = datetime.datetime.now()

formatted = now.strftime("%d-%m-%Y %H:%M:%S %A")
print("Formatted:", formatted)
```

<br>

#### VI) Convert String → Date (strptime)


```python
from datetime import datetime

date_str = "30-10-2025 14:45:00"
date_obj = datetime.strptime(date_str, "%d-%m-%Y %H:%M:%S")

print("Date Object:", date_obj)
```

# 2. 📅 Python calendar Module



#### I) Display a Month Calendar

```python
import calendar

print(calendar.month(2025, 10))
```

#### II) Display Full Year Calendar

```python
import calendar

print(calendar.calendar(2025))
```

#### III) Get Weekday of a Given Date

```python
import calendar

day = calendar.weekday(2025, 10, 30)
print(day)
```

#### IV) Check for Leap Year

```python
import calendar

print(calendar.isleap(2024))  # True
print(calendar.isleap(2025))  # False
```

#### V) Count Leap Years Between Two Years

```python
import calendar

print(calendar.leapdays(2000, 2025))  # Counts leap years between 2000 and 2024
```

#### VI) Get First Weekday and Total Days in a Month

```python
import calendar

month_info = calendar.monthrange(2025, 11)
print(month_info)
```
